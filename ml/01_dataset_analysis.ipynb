```python
df["Avg_Light"] = df[
    ["S1_Light", "S2_Light", "S3_Light", "S4_Light"]
].mean(axis=1)
```

```python
df["PIR_Motion"] = (
    (df["S6_PIR"] == 1) |
    (df["S7_PIR"] == 1)
).astype(int)
```

```python
df["Occupied"] = (
    df["Room_Occupancy_Count"] > 0
).astype(int)
```

```python
from sklearn.model_selection import train_test_split
```

```python
X = ml_data[
    [
        "Avg_Temperature",
        "Avg_Light",
        "PIR_Motion"
    ]
]

y = ml_data["Occupied"]
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)
```

```python
print("X_train:", X_train.shape)
print("X_test :", X_test.shape)
print("y_train:", y_train.shape)
print("y_test :", y_test.shape)
```

```python
print("Training distribution:")
print(y_train.value_counts())

print("\nTesting distribution:")
print(y_test.value_counts())
```

```python
## Now: Train Our First ML Model
```

```python
## Import Logistic Regression
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
model_lr = LogisticRegression(
    random_state=42,
    max_iter=1000
)
```

```python
model_lr.fit(X_train, y_train)
```

```python
y_pred_lr = model_lr.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

```python
print("Accuracy:", accuracy_score(y_test, y_pred_lr))
```

```python
print(classification_report(y_test, y_pred_lr))
```

```python
#Confusion Matrix
```

```python
cm_lr = confusion_matrix(y_test, y_pred_lr)

print(cm_lr)
```

```python
# Import Decision Tree
```

```python
from sklearn.tree import DecisionTreeClassifier

model_dt = DecisionTreeClassifier(
    random_state=42,
    max_depth=5
)

model_dt.fit(X_train, y_train)

y_pred_dt = model_dt.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

print("Decision Tree Accuracy:", accuracy_score(y_test, y_pred_dt))

print("\nClassification Report:")
print(classification_report(y_test, y_pred_dt))

print("\nConfusion Matrix:")
cm_dt = confusion_matrix(y_test, y_pred_dt)
print(cm_dt)
```

```python
from sklearn.tree import plot_tree
import matplotlib.pyplot as plt

plt.figure(figsize=(20, 10))

plot_tree(
    model_dt,
    feature_names=X.columns,
    class_names=["Empty", "Occupied"],
    filled=True,
    rounded=True
)

plt.title("Decision Tree for Room Occupancy Detection")
plt.show()
```

```python
# KNN
```

```python
from sklearn.neighbors import KNeighborsClassifier

model_knn = KNeighborsClassifier(
    n_neighbors=5
)

model_knn.fit(X_train, y_train)

y_pred_knn = model_knn.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

print("KNN Accuracy:", accuracy_score(y_test, y_pred_knn))

print("\nClassification Report:")
print(classification_report(y_test, y_pred_knn))

print("\nConfusion Matrix:")
cm_knn = confusion_matrix(y_test, y_pred_knn)
print(cm_knn)
```

```python
# Random Forest
```

```python
from sklearn.ensemble import RandomForestClassifier

model_rf = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

model_rf.fit(X_train, y_train)

y_pred_rf = model_rf.predict(X_test)
```

```python
# Evaluate Random Forest
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

print("Random Forest Accuracy:", accuracy_score(y_test, y_pred_rf))

print("\nClassification Report:")
print(classification_report(y_test, y_pred_rf))

print("\nConfusion Matrix:")
cm_rf = confusion_matrix(y_test, y_pred_rf)
print(cm_rf)
```

```python
# Random Forest Feature Importance 🌲

# This tells us which of your three sensor features contributes most to occupancy prediction.

import pandas as pd
import matplotlib.pyplot as plt

feature_importance = pd.DataFrame({
    "Feature": X.columns,
    "Importance": model_rf.feature_importances_
})

feature_importance = feature_importance.sort_values(
    by="Importance",
    ascending=False
)

print(feature_importance)
```

```python
# Visualize it

plt.figure(figsize=(8, 5))

plt.bar(
    feature_importance["Feature"],
    feature_importance["Importance"]
)

plt.xlabel("Features")
plt.ylabel("Importance")
plt.title("Random Forest Feature Importance")
plt.xticks(rotation=20)
plt.show()
```

```python
# Save the Random Forest model
import joblib

joblib.dump(model_rf, "../ml/occupancy_model.pkl")

print("Random Forest model saved successfully!")
```
