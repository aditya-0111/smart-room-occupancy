```python

# Create binary occupancy target
df["Occupied"] = (
    df["Room_Occupancy_Count"] > 0
).astype(int)

# Use one sensor reading per feature (no averaging)
ml_data = df[
    [
        "S1_Temp",
        "S1_Light",
        "S6_PIR",
        "Occupied"
    ]
].copy()

# Rename columns to match our hardware and backend
ml_data.columns = [
    "Temperature",
    "Light",
    "PIR_Motion",
    "Occupied"
]

# Convert training light values to binary states
ml_data["Light"] = (ml_data["Light"] > 0).astype(int)

# Save the updated dataset
ml_data.to_csv(
    "../data/occupancy_ml_dataset.csv",
    index=False
)

print(ml_data.head())
print("Dataset shape:", ml_data.shape)
```

```python
 print(ml_data["Light"].value_counts().sort_index())
```

```python
#Check the sensor inputs and target
print("Temperature summary:")
print(ml_data["Temperature"].describe())

print("\nLight summary:")
print(ml_data["Light"].describe())

print("\nPIR motion distribution:")
print(ml_data["PIR_Motion"].value_counts())

print("\nOccupancy distribution:")
print(ml_data["Occupied"].value_counts())
```

```python
print("Unique Light values:")
print(sorted(ml_data["Light"].unique()))

print("\nLight value counts:")
print(ml_data["Light"].value_counts().sort_index())
```

```python
#Select features and target
from sklearn.model_selection import train_test_split

X = ml_data[
    ["Temperature", "Light", "PIR_Motion"]
]

y = ml_data["Occupied"]

print("Features:", X.columns.tolist())
print("Target:", y.name)
```

```python
#Split training and testing data
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("X_train:", X_train.shape)
print("X_test:", X_test.shape)
print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
```

```python
#Check class distribution
print("Training distribution:")
print(y_train.value_counts())

print("\nTesting distribution:")
print(y_test.value_counts())
```

```python
#Train the Random Forest model
from sklearn.ensemble import RandomForestClassifier

model_rf = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

model_rf.fit(X_train, y_train)

print("Random Forest training completed.")
```

```python
#Evaluate the model
from sklearn.metrics import (
    accuracy_score,
    confusion_matrix,
    classification_report
)

y_pred_rf = model_rf.predict(X_test)

print("Accuracy:", accuracy_score(y_test, y_pred_rf))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred_rf))

print("\nClassification Report:")
print(classification_report(y_test, y_pred_rf))
```

```python
#Check feature importance
import pandas as pd

feature_importance = pd.Series(
    model_rf.feature_importances_,
    index=X.columns
).sort_values(ascending=False)

print(feature_importance)
```

```python
#Save the updated model
import joblib

joblib.dump(
    model_rf,
    "../ml/occupancy_model.pkl"
)

print("Updated Random Forest model saved successfully.")
```
